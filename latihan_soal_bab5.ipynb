### 6. Asumsikan bahwa daftar nums = [7, 3, 9, 1], dan tentukan output yang ditampilkan oleh baris kode.

```python
nums = [7, 3, 9, 1]

# a
print("Largest Number:", max(nums))

# b
print("Length:", len(nums))

# c
print("Total:", sum(nums))

# d
print("Number lot", sum(nums) / len(nums))
```
