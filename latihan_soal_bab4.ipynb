## Soal 3 - Persentase Mahasiswa Fakultas Vokasi ITS

```python
print("{0:20s}{1:s}".format("Departemen", "Persentase Mahasiswa"))
print("{0:14s}{1:10.1%}".format("Statistika Bisnis", .20))
print("{0:14s}{1:10.1%}".format("Manajemen Bisnis", .25))
print("{0:14s}{1:10.1%}".format("Teknik Kelautan", .15))
print("{0:14s}{1:10.1%}".format("Teknik Sipil", .20))
print("{0:14s}{1:10.1%}".format("Teknik Mesin", .20))
```

## Soal 4 - Harga Setelah Diskon

```python
harga_asli = 150000
diskon = 0.10
harga_diskon = harga_asli * (1 - diskon)

print("Harga Asli\t\t= Rp. {:,.0f}".format(harga_asli))
print("Diskon\t\t\t= {:.0%}".format(diskon))
print("Harga setelah diskon\t= Rp. {:,.0f}".format(harga_diskon))
```

## Soal 5 - Laba Bersih Perusahaan

```python
pendapatan = int(input("Masukkan Pendapatan: "))
pengeluaran = int(input("Masukkan Pengeluaran: "))
laba_bersih = pendapatan - pengeluaran

print("Pendapatan\t= Rp. {:,.0f}".format(pendapatan))
print("Pengeluaran\t= Rp. {:,.0f}".format(pengeluaran))
print("Laba Bersih\t= Rp. {:,.0f}".format(laba_bersih))
```
